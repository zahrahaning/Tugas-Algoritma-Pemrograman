```r
cek_bilangan <- function(x, y) {
    hasil <- x * y

    if (hasil %% 2 == 0) {
        return("genap")
    } else {
        return("ganjil")
    }
}

# Uji coba beberapa bilangan
cek_bilangan(2, 4)
cek_bilangan(1, 9)
cek_bilangan(13, 14)
```

```r
cek_bilangan <- function(x, y) {
    hasil <- x / y

    if (hasil %% 2 == 0) {
        return("genap")
    } else {
        return("ganjil")
    }
}

# Uji coba beberapa bilangan
cek_bilangan(4, 2)
cek_bilangan(9, 1)
cek_bilangan(182, 14)
```
